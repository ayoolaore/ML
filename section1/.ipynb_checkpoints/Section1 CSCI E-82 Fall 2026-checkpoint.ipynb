## Tip: efficient, reproducible environment management

You may have heard the saying that data science is 80 percent data cleaning. While that can be true at times, you may also find that environment setup and management can be a significant chunk of the time for any project if you face challenges seting up your environment. Since each person's local environment can be different due to the time a notebook was first built, as well as differences between systems themselves, it's useful to learn the skill of how to do this. 

With the above commands, you can stand up bespoke virtual environments for use in a single notebook very easily. By copying code from one notebook to another, you can reproduce the environment you had for a given project. However, as you gain more experience with Python and Linux, you will likely find yourself reusing most of the same code over and over again. Moreover, one of the **key benefits of installing from conda is that it works through dependencies between packages for you (most of the time)**. You will see this best if you set up your environment from the command line. Notice how it spends a lot of time solving for the environment before completing the installation. If you do enough environment installations, you'll also begin to notice that sometimes you don't seem to always get to a successful installation using the same command. There are at least two reasons for this: (1) order of the installation - if installing packages in series, one after another, you may be setting up dependencies as you go that are not satisfiable; (2) lack of specificity with a constraining dependency - an environment might only be solvable if you specify a specific version. 

The best method to remedy these types of issues is to build and test your environments more systematically than doing this one command at a time. We demonstrate a framework for this that will enable you to very rapid environment setup and debugging. This technique is especially useful for managing GPU aware installations on Linux, for example. Typical steps:

(1) **Identify the packages** you think you'll need - you probably won't know this the first time through a notebook. It's very common to realize that you want some extensibility during a project and to add it then;

(2) **Construct an environment file** that contains all of the dependencies you need through your notebook - over time, I have worked out a system for this and I reuse, create new versions, and repurpose environments across different projects so that I manage that knowledge effectively;

(3) **Decide on success criterion for your environment** - always, you want the packages to run. For GPU aware versions of your project, an additional success criterion is that you would want your libraries to be able to process commands through a GPU so that the notebook runs much faster - essential for serious, iterative deep learning;

(4) **Run the environment setup** - you will likely find that, for certain packages to work, you will need to fine tune this file a bit to satisfy all the dependencies you need. Handling this up front, and circling back as you run through your notebook when you find a new package requirement will save you time and frustration later. 

(5) **Test the environment** - typically done by running the notebook. For example, run commands from the library you're working with to verify that the new environment is GPU aware. Doing so up front before doing anything else has a good return on time since CPU driven deep learning takes quite a bit longer to run.

Reference: https://docs.conda.io/projects/conda/en/latest/user-guide/tasks/manage-environments.html

## How to write a clean notebook

Notebooks are more than a place to write and run code. They're a way to communicate your ideas and iterate on experiments so that someone else can understand what you're doing and follow along.

- Show your work and the results (be generous with comments)
- Write clear explanations of what you tried and why (if you tried something that didn't work, tell us why you think it didn't work and what you're going to try next)
- Avoid extra long lines of code (these often get cropped out when you export to PDF)
- Try to write code in a linear way (imagine someone reading your notebook from top to bottom, can they follow along?)
- Tools are sometimes too verberse in the output they produce, which clogs up a notebook and makes it hard to follow along. Try to filter this out.
- Make sure all datasets and other assets are included in your submission. Graders should be able to open a notebook, hit run, and have everything just work.

Here's an example of a **code block which isn't very clear:**

````python
def find(start, end): # Method lacks a clear name and description, and doesn't show the iterations you took to get here
    primes = []
    for x in range(start, end):
        print(f"Is {x} prime?") # Excess debug statements
        if x < 2:
            print(f"{x} is not prime")
            continue
        is_prime = True
        for y in range(2, x - 1):
            if x % y == 0:
                print(f"{x} % {y} == {x % y}") # Unclear what code is doing
                is_prime = False
                break
        if is_prime:
            print(f"{x} is prime")
            primes.append(x)
        else:
            print(f"{x} is not prime")
            
    return primes

primes = find(0, 20) # Results not printed
````

And here's a **better version** of the same code:

````python
import time

# Attempt 1: Generate an array of prime numbers by testing all
# numbers between 1 and each candidate number (exclusively)
def find_primes_exhaustively(start, end):
    primes = []
    for x in range(start, end):

        # The only even prime number is 2, so we'll make this a
        # special case
        if x < 2:
            continue

        # Assume each number is prime initially
        is_prime = True

        # Prime numbers are only divisible by 1 and themselves, so
        # check all numbers within that range
        for y in range(2, x - 1):

            # If the number is evenly divisible, it's not prime
            if x % y == 0:
                is_prime = False
                break

        # Add the number to the list of primes if it's prime
        if is_prime:
            primes.append(x)
    
    return primes

# Generate an array of primes between 0 and 1000
start_time = time.time()
primes = find_primes_exhaustively(0, 1000)
end_time = time.time()

# Show the results
print(primes)
print(f"Found {len(primes)} primes in {end_time - start_time} seconds")

# This code works but generates primes fairly inefficiently. Let's try an alternative algorithm which
# takes advantage of the fact that we only need to evaluate candidate numbers against the existing
# set of prime numbers that are less than the candidate we are testing.
````

````python
# Attempt 2: Generate an array of prime numbers more efficiently by
# only checking against previously found prime numbers
def find_primes_efficiently(end):
    primes = []
    for x in range(end):

        # There are no primes smaller than 2
        if x < 2:
            continue

        # The only even prime number is 2, so we'll use this as our
        # seed prime for finding other primes
        if x == 2:
            primes.append(x)
            continue

        # Assume each number is prime initially
        is_prime = True

        # Is a number is not divisible by any prime number smaller 
        # than it is, it is prime
        for y in primes:

            # If the number is evenly divisible, it's not prime
            if x % y == 0:
                is_prime = False
                break

        # Add the number to the list of primes if it's prime
        if is_prime:
            primes.append(x)
    
    return primes

# Generate an array of primes up to 1000
start_time = time.time()
primes = find_primes_efficiently(1000)
end_time = time.time()

# Show the results
print(primes)
print(f"Found {len(primes)} primes in {end_time - start_time} seconds")
````

This example is perhaps a bit heavy on the comments. You don't have to add comments to this extent, but it's better to err on the side of more comments since it demonstrates that you know what you're doing and requires less interpretation on the part of the person reading your notebook.

## Principal Component Analysis (PCA)

Why PCA:
- Reduce dimensionality of the dataset with minimal loss of information.
- Compress data without much loss of information, e.g., image compression
- As a first step in Feature engineering, Principal Component Regression, to quick rank observations,...

Principal Components:
- A subset of ordered abstract projection that can best **preserve the variations and correlations** of the original data
- A dataset with $n$ instances and $p$ features ($p$-dimensional data) can be approximately reproduced using $k < p$ Principal Components 

Mathematically:
- An eigendecomposition of the covariance matrix
- Finds directions that maximize the variance in the dataset.
- In PCA, we are projecting the entire set of data onto a different subspace
- In PCA, we are trying to find the axes with maximum variances where the data is most spread


PCA can be done by eigenvalue decomposition of a data covariance (or correlation) matrix or singular value decomposition of a data matrix, usually after a normalization step of the initial data. [Difference between eigen decomposition and SVD decomposition](https://math.stackexchange.com/questions/320220/intuitively-what-is-the-difference-between-eigendecomposition-and-singular-valu)

We might be interested in keeping only those eigenvectors with the largest eigenvalues, since they contain more information about the data. 

Create new feature subspace:
- eigenvalues that are close to 0 are less informative: they are dropped when we construct new feature subspace.

Algorithms for PCA
- Load a dataset that consists of $p$ features
- Compute the covariance matrix of the data
- Compute the eigenvectors and the corresponding eigenvalues
- Sort the eigenvectors by decreasing eigenvalues
- Select the top $k$ eigenvectors with the largest eigenvalues to form $p \times k$ matrix
- Use the $p \times k$ eigenvector matrix to transform the sample to a new subspace.

Step-by-step Examples
- PCA (with no sklearn)
- US Arrests data (with and without sklearn)

### Step 1: Create and plot the raw data

````python
import numpy as np
x = [-0.4, -0.3, 0.3, 2.5,0.5, 2.2, 1.9, 3.1, 2.3, 2, 1, 1.5, 1.1]
y = [-0.7,-0.4, 0., 2.4, 0.7, 2.9, 2.2,3.0,2.7,1.6,1.1,1.6,0.9]
data = np.array([x,y]).T

data
````

````python
import pandas as pd
df = pd.DataFrame(data, columns=['x', 'y'])
print(df.shape)
df.head()
````

````python
# This is just a helper function to get a plot which is centered on the origin in Matplotlib

set2_colors = ['#66c2a5','#fc8d62','#8da0cb','#e78ac3','#a6d854','#ffd92f','#e5c494','#b3b3b3']
light_grey = np.array([float(248)/float(255)]*3)
shade_black = '#262626'

#What are spines, ticks, labels, etc - https://matplotlib.org/3.1.1/gallery/showcase/anatomy.html

def cible_border(axes=None, top=False, right=False, left=True, bottom=True):
    """
    Make a target axis at 0,0 with ticks along the axis lines
    
    The top/right/left/bottom keywords toggle whether the corresponding plot border is drawn
    """
    ax = axes or plt.gca()

    ax.spines['right'].set_color('none')
    ax.spines['top'].set_color('none')
    ax.xaxis.set_ticks_position('bottom')
    ax.spines['bottom'].set_position(('data',0))
    ax.yaxis.set_ticks_position('left')
    ax.spines['left'].set_position(('data',0))

    #now re-enable visibles
    if top:
        ax.xaxis.tick_top()
    if bottom:
        ax.xaxis.tick_bottom()
    if left:
        ax.yaxis.tick_left()
    if right:
        ax.yaxis.tick_right()
````

````python
import matplotlib.pyplot as plt
set2_colors = ['#66c2a5','#fc8d62','#8da0cb','#e78ac3','#a6d854','#ffd92f','#e5c494','#b3b3b3']
ax = plt.gca()
ax.axis([-3, 4, -3, 4])
ax.scatter(df.x, df.y, s=100, c=set2_colors[0])
ax.set_title('Raw Data')
ax.grid(True)
cible_border()
plt.show()
````

### Step 2: Subtract the mean (center the data)

````python
mu = df.mean(axis=0)
print ("Mean:\n", mu, "\n")

df_0 = df - mu

ax = plt.gca()
ax.axis([-3, 4, -3, 4])
ax.scatter(df.x, df.y, s=100, c=set2_colors[0], label = "Raw Data")
ax.scatter(df_0.x, df_0.y, s=100, c=set2_colors[1], label = "Centered Data")
ax.set_title('Raw and mean-subtracted, centered data')
ax.grid(True)
ax.legend()
cible_border()

plt.show()
````

### Step 3: Calculate the covariance matrix $\Sigma$

$ var(x)= \frac{ \sum_{i=1}^{n} (X_i - \overline{X}) (X_i - \overline{X}) }{n-1} $  <br/>
(numpy.var default is divide by N)

<br/> <br/>
$ cov(X,Y)= \frac{ \sum_{i=1}^{n} (X_i - \overline{X}) (Y_i - \overline{Y}) }{n-1}  $
(numpy.cov default is divide by N-1, sample covariance, unbiased estimate) 

$
\sum = \begin{pmatrix}
var(x) & cov(x,y)\\ 
 cov(x,y)& var(y) 
\end{pmatrix}
$

````python
#D = np.cov(df.x,df.y)  # all of the above gives the following output
#print(D)

#C = np.cov(df.T)  #OR 
#C = np.cov(df, rowvar=0) #OR 
C = np.cov(df_0.x,df_0.y)  # all of the above gives the following output
#C is a positive semi definite matrix (A positive semi definite matrix is a symmetric matrix with all positive >=0 eigenvalues)
print(C)
````

- The diagonal (top left to bottom right) of covariance matrix is just variance cov(x,x) ~ 1.212, cov(y,y) ~ 1.55
- verify that with this command `np.var(df_0.x,ddof=1) `
- cov(x,y) ~1.333  = Since this is positive, both x and y are increasing/decreasing together. 
(This can be negative when x increases and y decreases or vice versa) 
- Magnitude of covariance matrix is not interpretable (unlike correlation coefficient), since it depends on data.  

### Step 4: Calculate the eigenvector and eigenvalues of the covariance matrix


````python
eigenvalues, eigenvectors = np.linalg.eigh(C)
print('Eigenvalues \n%s\n' %eigenvalues)
print('Eigenvectors \n%s' %eigenvectors)
````

####  The eigenvectors are all orthonormal 
(Dot product is zero and vectors are unit length)

````python
print(np.dot(eigenvectors[1], eigenvectors[0])) #Two vectors are orthogonal if the dot product of two vectors is 0
print('')
for vec in eigenvectors:
    print(np.round(np.linalg.norm(vec),6))
    
````

#### Sort the eigenvalues in descending order 
- This gives you the components in order of significance. 

````python
idx = np.argsort(-eigenvalues )
eigenvalues = eigenvalues [ idx ]
eigenvectors = eigenvectors [: , idx ]
print(eigenvalues)
print(idx)
print('')
print(eigenvectors) ## columns  also called phi1 , phi2,  or  PC1, PC2 
print('')
````

#### Check the the SVD decomposition

$A = USV^{T}$ 

[np.linalg.svd](https://docs.scipy.org/doc/numpy/reference/generated/numpy.linalg.svd.html)

## A look at real data - U.S. Arrests dataset

````python
import sklearn
from sklearn import decomposition, preprocessing
from mpl_toolkits.mplot3d import Axes3D
import matplotlib as mpl
import matplotlib.pyplot as plt
````

````python
#Originally this dataset is part of ISLR book/package in R - See chapter 10 in ISLR book
url = 'http://vincentarelbundock.github.io/Rdatasets/csv/datasets/USArrests.csv'   
usarrests = pd.read_csv(url,sep=',')
# https://stackoverflow.com/questions/11346283/renaming-column-names-in-pandas
usarrests = usarrests.rename(columns={'rownames':'State'})
usarrests.head()
````

````python
usarrests_scaled = usarrests.copy()
cols = usarrests_scaled.columns[1:]
usarrests_scaled[cols] = preprocessing.StandardScaler().fit_transform(usarrests_scaled[cols])
usarrests_scaled.head()
````

````python
pca = sklearn.decomposition.PCA(n_components=4).fit(usarrests_scaled.iloc[:,1:])

coef_PCA = pca.transform(usarrests_scaled.iloc[:,1:]) #coef_PCA = projected Data/ PC scores
# we make a scree plot to see how many Principal Components to consider
plt.figure(figsize=(12, 6))
eig = pca.explained_variance_
# and calculate the variance explained by the PC analysis
var_exp = pca.explained_variance_ratio_.cumsum()*100.

plt.plot(np.arange(1,len(eig)+1), eig, color='r')
plt.title('Scree plot for the PCA')
plt.xlabel('Number of principal components')
plt.ylabel('Eigenvalues')
plt.show()

print ('The 1st Principal Component explains {:03.1f} % of the variance\n'.format(var_exp[0]))
print ('The 1st and 2nd Principal Components explain {:03.1f} % of the variance\n'.format(var_exp[1]))
print ('The 1st, 2nd and 3rd Principal Components explain {:03.1f} % of the variance\n'.format(var_exp[2]))
print ('The first four Principal Components explain {:03.1f} % of the variance\n'.format(var_exp[3]))
````

### Feature loading plot
We can also plot the principal component loading vector coefficient for each of the features. This allows us to make an interpretation of the Principal Components on the basis of the dispersion of the feature loadings.

This is the a way of showing graphically the Correlation Circle that was referred to in lecture. ((See http://www.sthda.com/english/articles/31-principal-component-methods-in-r-practical-guide/112-pca-principal-component-analysis-essentials/ for correlation circle interpretation)

````python
def make_loading_plot(components, cols):
    summ_up = lambda x, y: abs(x) + abs(y)
    loadings = pd.DataFrame(data=dict(X=components[0], Y=components[1], name=cols))
    #loadings['summ'] = loadings.apply(lambda row: summ_up(row['X'], row['Y']), axis=1)
    plt.scatter(loadings.X, loadings.Y, s=200, label="Features", 
                facecolor=dark2_colors[1], edgecolor=shade_black, linewidth=0.35);

    for i in range(loadings.shape[0]):
        plt.text(loadings.iloc[i]['X']+.01, loadings.iloc[i]['Y'], loadings.iloc[i]['name'], size=12, color=shade_black)  

    plt.title("Feature loadings for the PCA");
    plt.xlabel("Feature loadings on PC1");
    plt.ylabel("Feature loadings on PC2");
    plt.grid(True)
    
    legend = plt.legend(loc=2,frameon=True,framealpha=0.6, scatterpoints=1)
    rect = legend.get_frame()
    rect.set_facecolor(light_grey)
    rect.set_linewidth(0.6)

make_loading_plot(pca.components_, usarrests_scaled.columns[1:])

plt.show();
````

## Perform PCA calculation

### Question:

*How many Principal Components can I have, maximum? *

Answer: 400 - because in this example we are treating the individual images as features, and the 4096 points as the data samples.  We can have a maximum of principal components equal to the number of features - i.e. the number of original images.   

````python
def pca_calculation(I, num_components):
    [nv , d] = I.shape
    print(nv,d)
    mean_I = np.mean(I, 0)
    ma_I = I - mean_I
    # calculate the eigenvectors (eigenimages) and eigenvalues
    M = np.dot(ma_I,ma_I.T)  #covariance matrix
    eigenvalues,eigenvectors = np.linalg.eigh(M)
    eigenvectors = np.dot(ma_I.T, eigenvectors)
    for i in range (nv):
        eigenvectors [:,i] = eigenvectors [:,i]/ np.linalg.norm ( eigenvectors [:, i])
    idx = np.argsort(-eigenvalues) #returns the index with increasing order 
                                   #(-z1 makes it decreasing order)
    eigenvalues = eigenvalues[idx]
    eigenvectors = eigenvectors[:, idx]
    
    print(M.shape[0:2])
    #return only num_components 
    eigenvalues = eigenvalues[0: num_components].copy()
    eigenvectors = eigenvectors[: ,0: num_components ].copy()
    return(eigenvalues, eigenvectors, mean_I) 


num_components = 8
eigenvalues, eigenvectors, mean_I = pca_calculation(I, num_components)
````

````python
eig = eigenvalues/np.sum(eigenvalues)
var_exp = eig.cumsum()*100.

plt.plot(np.arange(1,len(eig)+1), eig, color='r')
plt.title('Scree plot for the PCA')
plt.xlabel('Number of principal components')
plt.ylabel('Eigenvalues')
plt.show()

print ('The 1st Principal Component explains {:03.1f} % of the variance\n'.format(var_exp[0]))
print ('The 1st and 2nd Principal Components explain {:03.1f} % of the variance\n'.format(var_exp[1]))
print ('The 1st, 2nd and 3rd Principal Components explain {:03.1f} % of the variance\n'.format(var_exp[2]))
print ('The first four Principal Components explain {:03.1f} % of the variance\n'.format(var_exp[3]))
print ('The first five Principal Components explain {:03.1f} % of the variance\n'.format(var_exp[4]))
print ('The first six Principal Components explain {:03.1f} % of the variance\n'.format(var_exp[5]))
````

````python
plt.figure(figsize=(4,4))
plt.axis('off')
plt.gray()
_ = plt.imshow(mean_I.reshape(m,n))
````

````python
def pca_reconstruction(image, numEvs):
    weights = np.dot( I[image].reshape(1,-1) - mean_I, eigenvectors[: ,0: numEvs ])
    #reconstruct using principal components
    recon = mean_I + np.dot( weights, eigenvectors[: ,0: numEvs ].T)
    plt.figure(figsize=(4,4))
    plt.axis('off')
    plt.imshow(recon.reshape(m,n))
    plt.show()
    
pca_reconstruction(30, 8)    
````

````python
num_components = 400
eigenvalues, eigenvectors, mean_I = pca_calculation(I, num_components)
pca_reconstruction(30, num_components)
````

#### Lambda functions 
Lambda functions are anonymous functions, typically small and throw-away functions, that have become popular in other scripting languages and are available in Python too.  They can be quite useful in some Python APIs for other software (e.g. Spark, Tensorflow), in particular when used with `filter()`, `map()` and `reduce()`.   
Syntax:  
```[returned_variable] = lambda argument_list: expression ```

````python
add_func = lambda x, y : x + y
add_func(3,4)
````

````python
#This is equivalent to above lambda function
def add_func(x,y):
    return x+y

add_func(3,4)
````

````python
mylist2 = [1,2,3,4,5,6,7,8,9]
mapped = map(lambda x:x+2, mylist2)
[i for i in mapped]
````

#### Copying - shallow or deep copy

````python
a = [[1, 2, 3], [4, 5, 6]]
b = a   # list(a) also returns a shallow copy
print (a, '\n', b)
````

````python
a[1][2] = 10
print (a, '\n', b)
````

When a deep copy is needed, use `deepcopy()` instead

https://docs.python.org/3/library/copy.html

````python
from copy import copy, deepcopy
a = [[1, 2, 3], [4, 5, 6]]
b = copy(a)    # copy() returns a shallow copy
c = deepcopy(a)   # deepcopy() returns a deep copy
a[1][2] = 10
print (a, b, c)
````

### Maxtrix multiplication and Vectorization - 
Multidimensional arrays allow us operations on data they contain with out the need to write loops such as for loops - a process known as vectorization
- replacing explicit loops with array expression 
- vectorized array operations are faster 

####  1D numpy array - elementwise operations

````python
a1 = np.array([2.5, 5., 3.2])
print("a1 = " , a1,'\n')
print("np.ones(3) = ",np.ones(3),'\n')
a2  = 2*np.ones(3)
print("a2 = 2*np.ones(3) = ", a2,'\n')

a3 = a1*a2  # element by element multiplication 
print("a3 = a1*a2 = " , a3, '\n')

a4 = 1/a1 # element-by-element division
print("a4 = 1/a1" , a4)
````

####  2D array - elementwise operations

````python
b1 = np.array([[1,2,3], [4,5,6], [6,7,8]])
print("b1 = \n", b1, "\n")

b2  = 2*b1
print("b2 = 2*b1 = \n",b2, "\n")

b3 = b1*b2  # element by element multiplication  
print("b3 = b1*b2 = \n", b3 , "\n") 

b4 = 1/b1 # element-by-element division  
print("b4 = 1/b1 = \n" ,b4,"\n") 

b5 = b1 ** 2 #elementwise squaring
print("b5 = b1 ** 2 = \n" ,b5 ,"\n")
````

For 2-D arrays it is equivalent to matrix multiplication, and for 1-D arrays to inner product of vectors

### Timing your algorithms

You can use the `time` and  `timeit` modules to calculate the elapsed time of your algorithms. The best example is to show the speed increase between a python list comprehension (i.e. for loop) and a vectorized operation in Numpy. 

This example comes from this short online [book on Numpy](https://www.labri.fr/perso/nrougier/from-python-to-numpy/#id7)

````python
# make a simple Python function with addition in a list comprehension
def add_python(Z1,Z2):
    return [z1+z2 for (z1,z2) in zip(Z1,Z2)]

# make a simple Numpy function using the vectorized add method
def add_numpy(Z1,Z2):
    return np.add(Z1,Z2)
````

````python
import time 

Z1 = np.random.sample(1000)*100000
Z2 = np.random.sample(1000)*100000

starttime = time.time()
add_python(Z1, Z2)
endtime = time.time()
print("With the Python list comprehension it took {:03.9f} seconds\n".format(endtime-starttime))

starttime = time.time()
add_numpy(Z1, Z2)
endtime = time.time()
print("With the Numpy add method it took {:03.9f} seconds\n".format(endtime-starttime))
````

With the `timeit` magic in Jupyter, you pass the function to timeit as an argument - this is useful if you have several parts to time and you don't want to do the math to find the deltas.  

You can find details about timeit module [here](https://docs.python.org/2/library/timeit.html), but basically, the `%timeit` magic allows you the power of the module, but with a simplified call in Jupyter.

````python
%timeit (add_python(Z1, Z2))

%timeit (add_numpy(Z1, Z2))
````
